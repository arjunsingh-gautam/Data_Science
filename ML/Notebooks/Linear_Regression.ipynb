```python
x=dataset.loc[:,'Salary'].values
y=dataset.loc[:,'YearsExperience'].values
```

```python
x=x.reshape(-1,1)
y=y.reshape(-1,1)
x.shape
```

```python
sc=StandardScaler()
x=sc.fit_transform(x)
y=sc.fit_transform(y)
print(x,y)
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=1/3,random_state=42)
```

```python
x_test,y_test
```

```python
x_train,y_train
```

```python
class SimpleLinearRegression:
  def __init__(self):
    self.m=None
    self.b=None
  def fit(self,x,y):
    N=len(x)
    sum_x=np.sum(x)
    sum_y=np.sum(y)
    sum_xy=np.sum(x*y)
    sum_xx=np.sum(x*x)
    self.m=(N*sum_xy-sum_x*sum_y)/(N*sum_xx-sum_x**2)
    self.b=(sum_y-self.m*sum_x)/N

  # Define predict as a method of the class
  def predict(self,x):
    return self.m*x+self.b
```

```python
model=SimpleLinearRegression()
model.fit(x_train,y_train)
print(f"Slope:{model.m}")
print(f"Intercept:{model.b}")
```

```python
print(f"Line Equation:{model.m} x + {model.b}")
```

```python
from sklearn.metrics import mean_squared_error
y_pred=model.predict(x_test)
mse=mean_squared_error(y_test,y_pred) # Now you can use mean_squared_error
print(f"Mean Squared Error:{mse}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Sample Data
X = np.array([1, 2, 3, 4, 5])
Y = np.array([2, 2.5, 3.5, 4, 4.5])

# Calculate slope (m) and intercept (b)
N = len(X)
m = (N * np.sum(X * Y) - np.sum(X) * np.sum(Y)) / (N * np.sum(X*2) - (np.sum(X))*2)
b = (np.sum(Y) - m * np.sum(X)) / N

# Prediction Function
def predict(x):
    return m * x + b

# Plotting
plt.scatter(X, Y, color="blue", label="Actual Data")
plt.plot(X, predict(X), color="red", label=f"Regression Line: Y={round(m,2)}X+{round(b,2)}")
plt.xlabel("Hours Studied")
plt.ylabel("Score")
plt.legend()
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
import numpy as np

# Data
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)  # Reshape for sklearn
Y = np.array([2, 2.5, 3.5, 4, 4.5])

# Create Model and Fit Data
model = LinearRegression()
model.fit(X, Y)

# Predictions
Y_pred = model.predict(X)

# Print Slope and Intercept
print(f"Slope (m): {model.coef_[0]}")
print(f"Intercept (b): {model.intercept_}")

# Plot
import matplotlib.pyplot as plt

plt.scatter(X, Y, color="blue", label="Actual Data")
plt.plot(X, Y_pred, color="red", label="Regression Line")
plt.xlabel("Hours Studied")
plt.ylabel("Score")
plt.legend()
plt.show()
```
