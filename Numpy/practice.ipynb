What is numpy
- numpy is numerical python 
- It is library which help us perform numerical computation on arrays and matrices
- Why we need to learn numpy

- In ML the data is represented as vectors or matrices
    - Eg. input feature,output feature
    - We need to perform operation on this vectors like normalisation,encoding,multiplication during training of models or while preprocessing data
    - Numpy enables us to handle this computation efficiently using vectorised operation instead iterations
    - This significantly improves performance 

- How Broadcasting works? (yt)[]

- Main features:
    - array creationg
    - vectorise operations
    - broadcasting
    - performance
    - manipulating: data-type,reshaping
    - masking
    - Aggregation
    - concatenation
    - stacking
    - spliting


```python
import numpy as np
from time import time
```

```python
# Without numpy:
n=10000
l1=np.random.rand(n)
l2=np.random.rand(n)
l3=[]
# element by element multiplication
start_time=time()
for i,j in zip(l1,l2):
    l3.append(l1*l2)

end_time=time()
t1=end_time-start_time
print(f"Time taken for entire operation:{t1}")
    
    
```

```python
# with numpy
start_time=time()
l3=l1*l2
end_time=time()
t2=end_time-start_time
print(f"TIme take for entire operation:{t2}")
```

```python
# Performance analysis:
print(f"Performance increased by:{t1/t2}")
```

```python
text="balloonballoonbal"
from collections import defaultdict
hashmap=defaultdict(int)
for ch in text:
    hashmap[ch]+=1
print(hashmap)
l=[]
for c in "balon":
    if c not in ('l','o'):
        l.append(hashmap.get(c))
    else:
        l.append((hashmap.get(c))//2)
print(l)
print(min(l))
        
```
