## Concatenate Operation

```python
import numpy as np

a = np.array([1,2,3])
b = np.array([4,5,6])
c = np.concatenate((a,b),axis=0)


print(c)   # [1 2 3 4 5 6]

```

```python
a = np.array([[1,2],[3,4]])
b = np.array([[5,6]])
print(np.concatenate((a,b), axis=0)) # Row wise

# [[1 2]
#  [3 4]
#  [5 6]]
```

```python
a = np.array([[1,2],[3,4]])
b = np.array([[5,6],[7,8]])
print(np.concatenate((a,b), axis=1)) # column wise
print(np. concatenate((a,b),axis=0)) # row wise
# [[1 2 5 6]
#  [3 4 7 8]]
```

```python
a = np.array([[1,2],[3,4]])
b = np.array([[5,6,7]])
np.concatenate((a,b), axis=1)  # ValueError
```

## H-Stack

```python
a=np.array([1,2,3])
b=np.array([4,5,6])
print(np.hstack((a,b)))
print(np.concatenate((a,b),axis=0))
```

```python
c=np.array([[1,2],[3,4]])
d=np.array([[5,6],[7,8]])
print(np.hstack((c,d)))
print(np.concatenate((c,d),axis=1))
```

## V-Stack

```python
print(np.vstack((a,b)))
print(np.concatenate((a,b),axis=1))
```

```python
print(np.vstack((c,d)))
print(np.concatenate((c,d),axis=0))
```

## Split

```python
arr1=np.array([1,2,3,4,5,6])
print(np.split(arr1,2))
```

```python
print(np.split(arr1,[1,4]))
```

```python
arr2=np.random.rand(4,4)
print(arr2)
```

```python
print(np.split(arr2,2,axis=0)) #vsplit

print()
print(np.vsplit(arr2,2))
```

```python
print(np.split(arr2,2,axis=1)) #hsplit
```
