# **<span style="color:orange">Indexing in Python</span>**

## **Basic Indexing**
- used for accessing elements in an array
- slicing operation
```python
arr[index]
arr[start:stop:step]
```

````python
import numpy as np
arr = np.array([10, 20, 30, 40, 50])
````

````python
arr[1:4] = 99
print(arr)
````

## **Fancy Indexing**

- Selecting elements using index arrays
```python
arr[[i1,i2,i3]]
```

````python
print(arr[[0,4]])
````

## **Boolean Indexing**
- Selecting elements using true False condition

```python
arr[condition]
```

````python
mask=arr[arr>30]
mask
````

````python
arr=np.arange(1,21)
arr[(arr%2==0) & (arr%3==0)]=0
print(arr)
````
