###`Q-1:` Find the nearest element in the array to a given integer.
```
Ex:-
a=23 and array - [10 17 24 31 38 45 52 59].
Nearest element is 24
```
Hint: Read about this function `argmin()`

````python
# code here
import numpy as np
a=int(input("Enter a number"))
arr=np.array([10,17,24,31,38,45,52,59])
diff=abs((arr-a))
print(diff)
index_min=diff.argmin()
print(f"Nearest elemment to {a} in array is:{arr[index_min]}")
````

###`Q-4:` Your are given an array which is havig some nan value. You job is to fill those nan values with most common element in the array.
```
arr=np.array([[1,2,np.nan],[4,2,6],[np.nan,np.nan,5]])

```

````python
# code here
arr=np.array([[1,2,np.nan],[4,2,6],[np.nan,np.nan,5]])
print("Original array:")
print(arr)

# Flatten the array and remove NaN values
flat_arr = arr[~np.isnan(arr)]
print(flat_arr)

# Find the unique values and their counts
values, counts = np.unique(flat_arr, return_counts=True)
print(values,counts)
# Find the most common value
if len(values) > 0:
    most_common_value = values[np.argmax(counts)]
else:
    most_common_value = 0 # Default if no non-nan values

print(f"Most common non-NaN value: {most_common_value}")

# Replace NaN values with the most common value
arr[np.isnan(arr)] = most_common_value

print("Array after filling NaNs:")
print(arr)
````

### `Q-5:` Write a NumPy program

- to find the missing data in a given array. Return a boolean matrix.
- also try to fill those missing values with 0. For that, you can use `np.nan_to_num(a)`

```python
import numpy as np

np.array([[3, 2, np.nan, 1],
          [10, 12, 10, 9],
          [5, np.nan, 1, np.nan]])
```

````python
# code here
arr=np.array([[3,2,np.nan,1],[10,12,10,9],[5,np.nan,1,np.nan]])
print("Original array:")
print(arr)
print("Find the missing data of the said array:")
print(np.isnan(arr))
print("Fill the missing values with 0:")
print(np.nan_to_num(arr,nan=0))
````

###`Q-6:` Given two arrays, X and Y, construct the Cauchy matrix C.
`Cij =1/(xi - yj)`

http://en.wikipedia.org/wiki/Cauchy_matrix
```
x = numpy.array([1,2,3,4]).reshape((-1, 1)
y = numpy.array([5,6,7])
```

````python
# code here
import numpy as np
x = np.array([1,2,3,4]).reshape((-1, 1))
y = np.array([5,6,7])

c = 1.0 / (x - y) # Broadcasting
print("Cauchy Matrix (Cij = 1/(xi - yj)):")
print(c)
````
