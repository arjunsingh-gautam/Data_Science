## Numpy Array Attributes

```python
import numpy as np
```

```python
a1=np.array([[1,2,3],[4,5,6]])
```

```python
print(a1.ndim)
```

```python
print(a1.size)
```

```python
print(a1.dtype)
```

```python
print(a1.itemsize)
```

```python
print(a1.nbytes)
```

```python
print(a1.T)
```

```python
for x in a1.flat:
    print(x)
```
