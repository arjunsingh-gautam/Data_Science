## Vectorisation and Vectorised Operation in Numpy and it's Performance benefits:

### Non Vectorised Array Addition

```python
n=1000000
a=np.random.rand(n)
b=np.random.rand(n)
c=[]
start_time=time()
for k in range(n):
    c.append(a[k]+b[k])

end_time=time()
total_time_1=end_time-start_time
print(f"Time taken for entire operation:{total_time_1}")
```

### Vectorised Operation of Array Addition

```python
start_time=time()
d=a+b
end_time=time()
total_time_2=end_time-start_time
print(f"Time taken for entire operation:{total_time_2}")
```

```python
print(c==d)
```

```python
# Increase in Performance to vectorised operaion
print(f"Performance increased by:{total_time_1/total_time_2} times")
```

- This code demonstrates how slicing a numpy array does create a copy of instead create a view of the original array with adjusted shape and stride

```python
arr=np.array([1,2,3,4,5])
new_arr=arr[:4] # does not create of slice list only create a view to original array
print(f"Old array:{arr}")
print(f"New Array:{new_arr}")
new_arr[0]=99 # changes original arr also
print(f"Old array:{arr}")
print(f"New Array:{new_arr}")
print(f"ID of arr:{id(arr)}")
print(f"ID of new_arr:{id(new_arr)}")
print(f"ID of arr[0]:{id(arr[0])}") # Share same memory buffer
print(f"ID of new_arr[1]:{id(new_arr[0])}")
print(arr is new_arr)
```

- To prevent this windowing of same memory buffer we can forcefully create a copy using copy() method

```python
arr=np.array([1,2,3,4,5])
new_arr=arr[:4].copy() #  slice the list and create a separate copy of slice list
print(f"Old array:{arr}")
print(f"New Array:{new_arr}")
new_arr[0]=99 # does not change original array
print(f"Old array:{arr}")
print(f"New Array:{new_arr}")
print(f"ID of arr:{id(arr)}")
print(f"ID of new_arr:{id(new_arr)}")
print(f"ID of arr[0]:{id(arr[0])}") # Share same memory buffer
print(f"ID of new_arr[0]:{id(new_arr[0])}")
print(arr is new_arr)
```

```python
print("Memory address of arr:", arr.__array_interface__['data'])
print("Memory address of new_arr:", new_arr.__array_interface__['data'])
```

## Numpy Performance efficiency

```python
import numpy as np
import time

# Python list
L = list(range(1000000))
start = time.time()
[x**2 for x in L]
print("List time:", time.time() - start)

# NumPy array
A = np.arange(1000000)
start = time.time()
A**2
print("NumPy time:", time.time() - start)
```

## Memory Efficiency

```python
import sys
L = list(range(10))
A = np.arange(10)

print("List size:", sys.getsizeof(L))
print("NumPy array size:", A.nbytes)
```

## Vectorised Operation

```python
# List
a=[x+5 for x in L]
print(a)
# NumPy
b=A + 5   # way faster
print(b)
```
