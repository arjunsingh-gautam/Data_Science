# **<span style="#ac7920">System of Linear Equations</span>**

```python
import numpy as np
```

```python
# Finding solution to system of linear equations:
# Solution of  x+2y=3 and 4x+5y=6
A=np.array([[1,2],[4,5]])
b=np.array([3,6])
x=np.linalg.solve(A,b)
print(f"Solution of Ax=b is:{x}")
```
