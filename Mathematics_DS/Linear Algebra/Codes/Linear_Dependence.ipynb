# **Linear Dependent or Independent set of Vectors**

````python
import numpy as np

def take_vectors_from_user():
    """
    Takes a set of vectors from the user.
    Ensures all vectors have the same dimension.
    Returns a matrix where each vector is a column.
    """

    num_vectors = int(input("Enter number of vectors: "))
    dimension = int(input("Enter dimension of each vector: "))

    vectors = []

    for i in range(num_vectors):
        print(f"Enter components of vector {i+1} (space separated):")
        vec = list(map(float, input().split()))

        if len(vec) != dimension:
            raise ValueError("All vectors must have the same dimension!")

        vectors.append(vec)

    # Convert to matrix with vectors as columns
    return np.array(vectors).T


def check_linear_dependence(matrix):
    """
    Checks whether the given set of vectors is
    linearly dependent or independent using rank.
    """

    rank = np.linalg.matrix_rank(matrix)
    num_vectors = matrix.shape[1]

    if rank == num_vectors:
        return "Linearly Independent"
    else:
        return "Linearly Dependent"


def main():
    matrix = take_vectors_from_user()
    result = check_linear_dependence(matrix)

    print("\nResult:")
    print(result)

if __name__ == "__main__":
    main()


````



## 🔴 First-Principle Idea (before code)

From first principles:

* Vectors are **linearly independent** if **each vector adds a new direction**
* This happens **iff** the number of **independent directions** equals the number of vectors

In linear algebra:

* Independent directions = **rank of matrix**
* Number of vectors = **number of columns**

👉 **Key rule**

```
If rank == number of vectors  → linearly independent
If rank < number of vectors   → linearly dependent
```

Why this works:

* Rank counts **non-redundant vectors**
* If rank is smaller, some vectors are **linear combinations of others**

## 🟡 Dry Run Example 1: Linearly Independent Vectors

### Input

```python
vectors = [
    [1, 0],
    [0, 1]
]
```

---

#### Step 3: Compare

```text
rank = 2

```

✅ rank == number of vectors
➡ **Linearly Independent**

## 🟡 Dry Run Example 2: Linearly Dependent Vectors

### Input

```python
vectors = [
    [1, 2],
    [2, 4],
    [3, 6]

```

---

### Step-by-Step Dry Run

#### Step 1: Matrix formation

```text
matrix =
[ 1  2  3 ]
[ 2  4  6 ]
```

Notice:

* Column2 = 2 × Column1
* Column3 = 3 × Column1

---

#### Step 2: Rank calculation

```python
rank = np.linalg.matrix_rank(matrix)
```

* Rank = 1
* Only **one unique direction**

---

#### Step 3: Compare

```text
rank = 1
num_vectors = 3
```

❌ rank < number of vectors
➡ **Linearly Dependent**

---

## 🔵 Why Rank Works (First Principles)

From first principles:

* Rank = number of **linearly independent columns**
* Rank ignores **redundant information**
* If vectors are dependent → rank drops

Equivalent to solving:

```
a₁v₁ + a₂v₂ + ... + aₙvₙ = 0
```

If non-zero solution exists → rank deficiency → dependence

## ✅ Final First-Principle Takeaway

> **A set of vectors is linearly independent if and only if the number of independent directions (rank) equals the number of vectors.**

This is why:

* ML checks **rank of feature matrix**
* PCA removes **rank-deficient directions**
* Multicollinearity breaks regression stability
