### What is Pandas

Pandas is a fast, powerful, flexible and easy to use open source data analysis and manipulation tool,
built on top of the Python programming language.

https://pandas.pydata.org/about/index.html

### Pandas Series

A Pandas Series is like a column in a table. It is a 1-D array holding data of any type.

### Importing Pandas

```python
import numpy as np
import pandas as pd
```

### Series from lists

```python
# string
country = ['India','Pakistan','USA','Nepal','Srilanka'] # Python List

pd.Series(country) # convert list to Pandas Series
```

```python
students={2207:"Arjun",2281:"Ashish",1921:"Bhoumik",2911:"Anuj"} # Python Dict
s=pd.Series(students) # key:index of each of element and value:element values of series
print(s)
print(type(s))
```

```python
# integers
runs = [13,24,56,78,100]

runs_ser = pd.Series(runs)
print(runs_ser)
```

```python
# custom index
marks = [67,57,89,100]
subjects = ['maths','english','science','hindi']

pd.Series(marks,index=subjects)
```

```python
# setting a name
marks = pd.Series(marks,index=subjects,name='Nitish ke marks') # name act as title for column or label of entire column
marks
```

## Uses of Each Metadata of Pandas Series

1. index
- Index is used to access data element of the series using indexing
- Used to align data while performing operatons on data

```python
# Label based acces
print(marks['science'])
```

```python
for x in marks:
    print(x)
```

```python
for x in subjects:
    print(marks[x]) # label based access
```

```python
# In this code we will see why index help in data alignment
s1 = pd.Series([92, 23], index=['x', 'y'])
s2 = pd.Series([20, 63], index=['y', 'x'])
s3 = s1 + s2    # ==> x=92+63 and y=20+23
print(s3)
```

```python
# name is used during ploting: used as label/legend
from matplotlib import pyplot as plt
plt.bar(marks.index, marks.values, label=marks.name)
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.legend()
plt.show()
```

```python
import pandas as pd

s = pd.Series([10, 20, 30], index=['a','b','c'], name="Sales")

print("Index:", s.index)       # Index(['a','b','c'])
print("Values:", s.values)     # array([10,20,30])
print("Dtype:", s.dtype)       # int64
print("Shape:", s.shape)       # (3,)
print("Size:", s.size)         # 3
print("Name:", s.name)         # Sales
print("Empty?", s.empty)       # False
print("Unique?", s.is_unique)  # True
print("Monotonic increasing?", s.is_monotonic_increasing)  # True
print("Bytes used:", s.nbytes) # 24 (depends on system)
```
