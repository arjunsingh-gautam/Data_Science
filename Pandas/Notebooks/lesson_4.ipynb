## **2. Filling Rows or Columns having Missing values**

```python
df2=df.copy()
```

```python
# filling using mean of remaining values
df2["Insulin"].isnull().sum()
```

```python
df2['Insulin'].mean()
```

```python
df2=df2['Insulin'].fillna(df2['Insulin'].mean())
```

```python
df2
```

```python
df2.isnull().sum()
```

```python
df.iloc[42:93]=np.nan
```

```python
df.isnull().sum()
```

```python
df.iloc[42:93]=df.iloc[42:93].fillna('Missing')
```

```python
df.iloc[42:93]
```

```python
df.isnull().sum()
```

```python
# Creating Missing Values
df.loc[45:92,'Age']=np.nan
```

```python
df.isnull().sum()
```

```python
df.loc[45:92,'Age']=df.loc[45:92,'Age'].fillna(df['Age'].mode()[0])
```

```python
df.loc[45:92,'Age']
```
