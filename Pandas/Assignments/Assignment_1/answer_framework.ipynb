# ⚙️ Other Methods to Find Top Performers

### **1️⃣ CPA-Based Ranking (Efficiency-Focused)**

[
\text{CPA} = \frac{\text{Spend}}{\text{Sales}}
]
✅ **Pros**

* Reflects *cost efficiency* — lower CPA means better ROI.
* Prioritizes strategies delivering more conversions per dollar.

❌ **Cons**

* Can overrate small-sample strategies with very few sales (unstable CPA).
* Ignores overall scale (sales volume).

---

### **2️⃣ Weighted Efficiency (Sales × 1/CPA)**

[
\text{Weighted Score} = \text{Sales} \times \frac{1}{\text{CPA}}
]
✅ **Pros**

* Balances **scale and efficiency** simultaneously.
* Prevents bias toward large-spend or small-CPA strategies alone.

❌ **Cons**

* Needs both metrics clean and reliable (avoid zero-sales division).
* Slightly more complex to explain to non-analysts.

---

### **3️⃣ Composite Score (Normalized Metrics)**

[
\text{Score} = w_1 \times \text{Sales} + w_2 \times (1/\text{CPA}) + w_3 \times \text{CTR}
]
✅ **Pros**

* Multi-dimensional — considers volume, cost, and engagement.
* Used in advanced campaign optimization models.

❌ **Cons**

* Needs proper weight tuning (can introduce bias).
* Harder to communicate in simple dashboards.

---

### **4️⃣ ROI or ROAS-Based Ranking (Profit-Driven)**

If revenue per sale known:
[
\text{ROAS} = \frac{\text{Revenue}}{\text{Spend}}
]
✅ **Pros**

* True profitability indicator.
* Great for optimizing budget allocation.

❌ **Cons**

* Requires accurate revenue data (often unavailable in task datasets).

# 🧮 1️⃣ How Weighted CPA Works — Simple Example

| Strategy | Sales | Individual CPA ($) | Weighted Impact |
| :------- | ----: | -----------------: | --------------: |
| A        |    10 |                100 |  10×100 = 1,000 |
| B        |   100 |                 50 |  100×50 = 5,000 |
| C        |     5 |                200 |   5×200 = 1,000 |

[
\text{Weighted CPA} = \frac{1{,}000 + 5{,}000 + 1{,}000}{10 + 100 + 5} = \frac{7{,}000}{115} = 60.9
]

✅ So even though Strategy C has a high CPA (200), its low sales (5) don’t skew the overall result.
Weighted CPA = realistic, business-scaled efficiency.

---

# ⚖️ 2️⃣ Why Use Weighted CPA (and Not Simple Average)

| Approach               | What It Does               | Problem                                                                    |
| :--------------------- | :------------------------- | :------------------------------------------------------------------------- |
| **Simple Average CPA** | Takes the mean of all CPAs | Treats all campaigns equally, even if one has 1 sale and another 100 sales |
| **Weighted CPA** ✅     | Weights by number of Sales | Reflects *true cost efficiency* of the total conversions generated         |

💬 *Example Insight:*

> “If one strategy made 2 sales at ₹10 CPA and another made 200 sales at ₹20 CPA, a simple average (₹15) would understate the true cost. Weighted CPA (≈₹19.8) better represents the actual campaign performance.”
