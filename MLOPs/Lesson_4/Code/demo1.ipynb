```python
import mlflow
```

```python
exp=mlflow.set_experiment("Regression Experiment")
```

```python
print(f"Experiment ID:{exp.experiment_id}")
```

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import ElasticNet
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score
import mlflow
import mlflow.sklearn
df=pd.read_csv(r"D:\Desktop\Data_Science\MLOPs\data\winequality-red.csv")

def train_model(alpha, l1_ratio):
    train,test=train_test_split(df,test_size=0.3)
    X_train=train.drop('quality',axis=1)
    y_train=train['quality']
    X_test=test.drop('quality',axis=1)
    y_test=test['quality']
    with mlflow.start_run(experiment_id=exp.experiment_id,
                          run_name="Regression",
                          description="Performing Regression Analysis"):

        # Model building and fitting
        model = ElasticNet(alpha=alpha, l1_ratio=l1_ratio)

        model.fit(X_train, y_train)
        # Model Prediction
        y_pred = model.predict(X_test)
        # Model Evaluation
        rmse = np.sqrt(mean_squared_error(y_test, y_pred))
        mse = mean_squared_error(y_test, y_pred)
        mae = mean_absolute_error(y_test, y_pred)
        r2 = r2_score(y_test, y_pred)
        print("ElasticNet Model (alpha=%f, l1_ratio=%f):" % (alpha, l1_ratio))
        print("RMSE:", rmse)
        print("MAE:", mae)
        print("MSE:", mse)
        print("R2:", r2)

        # Logging
        mlflow.log_param("alpha", alpha)
        mlflow.log_param("l1_ratio", l1_ratio)
        mlflow.log_metric("rmse", rmse)
        mlflow.log_metric("mse", mse)
        mlflow.log_metric("mae", mae)
        mlflow.log_metric("r2", r2)
        mlflow.sklearn.log_model(model, "model")
```

```python
for alpha in [0.01,0.1,1]:
    for l1 in [0.1,0.5,0.9]
        train_model(alpha,l1)
```
